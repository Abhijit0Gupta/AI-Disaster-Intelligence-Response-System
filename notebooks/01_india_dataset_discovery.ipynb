# India Flood Dataset — Initial Discovery

This notebook performs the initial audit of the raw India flood dataset.

Important:
- The raw dataset is read-only.
- No modifications are made to the raw CSV.
- Existing cleaned/feature-engineered datasets are not used at this stage.
- No assumptions are made about the target or feature definitions.

```python
from pathlib import Path
import pandas as pd

# Project root
PROJECT_ROOT = Path.cwd().parent

# Raw India dataset
RAW_DATA_PATH = PROJECT_ROOT / "data" / "raw" / "India_Flood_Inventory_v3 (2).csv"

print("Project root:", PROJECT_ROOT)
print("Raw dataset path:", RAW_DATA_PATH)
print("File exists:", RAW_DATA_PATH.exists())

if not RAW_DATA_PATH.exists():
    raise FileNotFoundError(
        f"Raw India dataset not found at:\n{RAW_DATA_PATH}"
    )

# Load raw dataset
df_raw = pd.read_csv(RAW_DATA_PATH)

print("\nRaw India dataset loaded successfully.")
print("Shape:", df_raw.shape)

print("\nColumn names:")
for i, column in enumerate(df_raw.columns, start=1):
    print(f"{i:>3}. {column}")

print("\nData types:")
print(df_raw.dtypes)

print("\nFirst 5 rows:")
display(df_raw.head())
```

```python
# Complete schema and data-quality overview

schema_df = pd.DataFrame({
    "Column": df_raw.columns,
    "Data Type": df_raw.dtypes.astype(str).values,
    "Non-Null Count": df_raw.notna().sum().values,
    "Missing Count": df_raw.isna().sum().values,
    "Missing %": (df_raw.isna().mean() * 100).round(2).values,
    "Unique Values": df_raw.nunique(dropna=True).values
})

print("RAW INDIA DATASET — COMPLETE SCHEMA")
print("=" * 80)

display(schema_df)

print("\nDataset shape:", df_raw.shape)

print("\nTotal missing values:", df_raw.isna().sum().sum())
print(
    "Overall missing percentage:",
    round(df_raw.isna().mean().mean() * 100, 2),
    "%"
)

print("\nDuplicate rows:", df_raw.duplicated().sum())
```

```python
# Display each column with a few representative non-null values

print("REPRESENTATIVE VALUES BY COLUMN")
print("=" * 80)

for column in df_raw.columns:
    print(f"\n[{column}]")
    print("dtype:", df_raw[column].dtype)
    print("sample values:")
    print(df_raw[column].dropna().head(5).tolist())
```

```python
# Exact raw column-name inspection
# IMPORTANT: This does NOT modify df_raw.

print("EXACT RAW COLUMN NAMES")
print("=" * 100)

for i, column in enumerate(df_raw.columns, start=1):
    print(f"{i:>2}. repr = {repr(column)}")
    print(f"    length = {len(column)}")

print("\n" + "=" * 100)
print("COLUMNS CONTAINING 'damage', 'casual', 'fatal', 'injur', or 'displac'")
print("=" * 100)

keywords = ["damage", "casual", "fatal", "injur", "displac"]

for column in df_raw.columns:
    if any(keyword in str(column).lower() for keyword in keywords):
        print(repr(column))
```

```python
# Automatically identify potential impact/outcome columns
# based only on their actual column names.

impact_keywords = [
    "fatal",
    "injur",
    "displac",
    "damage",
    "casual",
    "affected",
    "death",
    "loss"
]

impact_candidates = [
    column
    for column in df_raw.columns
    if any(keyword in str(column).lower() for keyword in impact_keywords)
]

print("POTENTIAL IMPACT / OUTCOME COLUMNS")
print("=" * 100)

for i, column in enumerate(impact_candidates, start=1):
    series = df_raw[column]

    print(f"\n{i}. {repr(column)}")
    print(f"   dtype: {series.dtype}")
    print(f"   non-null: {series.notna().sum()}")
    print(f"   missing: {series.isna().sum()}")
    print(f"   missing %: {series.isna().mean() * 100:.2f}%")
    print(f"   unique non-null: {series.nunique(dropna=True)}")

    print("   sample values:")
    for value in series.dropna().astype(str).head(10):
        print(f"      {value}")
```

```python
# Compact value-profile for every raw column
# This is inspection only. No data is modified.

profile_rows = []

for column in df_raw.columns:
    series = df_raw[column]

    non_null = series.dropna()

    sample_values = non_null.astype(str).head(3).tolist()

    profile_rows.append({
        "Column": column,
        "Dtype": str(series.dtype),
        "Non_Null": series.notna().sum(),
        "Missing": series.isna().sum(),
        "Missing_%": round(series.isna().mean() * 100, 2),
        "Unique": series.nunique(dropna=True),
        "Sample_1": sample_values[0] if len(sample_values) > 0 else None,
        "Sample_2": sample_values[1] if len(sample_values) > 1 else None,
        "Sample_3": sample_values[2] if len(sample_values) > 2 else None
    })

column_profile = pd.DataFrame(profile_rows)

display(column_profile)
```

```python
# Numerical-column summary

numeric_columns = df_raw.select_dtypes(include="number").columns.tolist()

print("NUMERICAL COLUMNS")
print("=" * 100)
print(numeric_columns)

if numeric_columns:
    numerical_summary = df_raw[numeric_columns].describe().T

    numerical_summary["missing"] = df_raw[numeric_columns].isna().sum()
    numerical_summary["missing_%"] = (
        df_raw[numeric_columns].isna().mean() * 100
    ).round(2)

    display(numerical_summary)
else:
    print("No numerical columns found.")
```

```python
# Categorical / string-column cardinality summary

string_columns = df_raw.select_dtypes(include=["object", "string"]).columns.tolist()

categorical_summary = []

for column in string_columns:
    series = df_raw[column]

    categorical_summary.append({
        "Column": column,
        "Unique": series.nunique(dropna=True),
        "Missing": series.isna().sum(),
        "Missing_%": round(series.isna().mean() * 100, 2),
        "Most_Frequent": series.value_counts(dropna=True).index[0]
            if series.notna().any() else None,
        "Most_Frequent_Count": series.value_counts(dropna=True).iloc[0]
            if series.notna().any() else 0
    })

categorical_summary = pd.DataFrame(categorical_summary)

display(categorical_summary)
```

```python
# Investigate where High_Impact appears in the local repository.
# This is inspection only. No files are modified.

from pathlib import Path

print("SEARCHING LOCAL REPOSITORY FOR 'High_Impact'")
print("=" * 100)

search_root = PROJECT_ROOT

matches = []

for path in search_root.rglob("*"):
    if not path.is_file():
        continue

    # Skip virtual environments and Git internals
    if ".venv" in path.parts or ".git" in path.parts:
        continue

    # Only inspect relevant text/code/data files
    if path.suffix.lower() not in {
        ".py", ".ipynb", ".csv", ".txt", ".md"
    }:
        continue

    try:
        text = path.read_text(encoding="utf-8", errors="ignore")
    except Exception:
        continue

    if "High_Impact" in text:
        matches.append(path.relative_to(PROJECT_ROOT))

if matches:
    print("Files containing 'High_Impact':")
    for path in matches:
        print(" -", path)
else:
    print("No occurrences of 'High_Impact' found.")
```

```python
# Inspect existing India processed datasets.
# These files are being audited, not trusted.

CLEANED_PATH = (
    PROJECT_ROOT / "data" / "processed"
    / "India_Flood_Inventory_cleaned.csv"
)

FEATURES_PATH = (
    PROJECT_ROOT / "data" / "processed"
    / "India_Flood_Inventory_features.csv"
)

print("EXISTING INDIA PROCESSED FILES")
print("=" * 100)

for name, path in [
    ("Cleaned", CLEANED_PATH),
    ("Features", FEATURES_PATH)
]:
    print(f"\n{name} dataset")
    print("-" * 100)
    print("Path:", path)
    print("Exists:", path.exists())

    if path.exists():
        temp_df = pd.read_csv(path)

        print("Shape:", temp_df.shape)
        print("Columns:")
        for i, column in enumerate(temp_df.columns, start=1):
            print(f"  {i:>2}. {repr(column)}")

        print("\nFirst 3 rows:")
        display(temp_df.head(3))
```

```python
# Inspect High_Impact wherever it exists.
# We are NOT assuming that it is a valid target.

for name, path in [
    ("Cleaned", CLEANED_PATH),
    ("Features", FEATURES_PATH)
]:
    if not path.exists():
        continue

    temp_df = pd.read_csv(path)

    if "High_Impact" in temp_df.columns:
        print("\n" + "=" * 100)
        print(f"HIGH_IMPACT FOUND IN: {name}")
        print("=" * 100)

        target = temp_df["High_Impact"]

        print("dtype:", target.dtype)
        print("rows:", len(target))
        print("missing:", target.isna().sum())
        print("unique values:", target.nunique(dropna=True))

        print("\nValue counts:")
        print(target.value_counts(dropna=False).sort_index())

        print("\nSample values:")
        print(target.head(20).tolist())

    else:
        print(f"\nHigh_Impact is NOT present in {name} dataset.")
```

```python
from pathlib import Path
import pandas as pd
import numpy as np

# ============================================================
# RELOAD RAW INDIA DATASET FOR TARGET LEAKAGE AUDIT
# ============================================================

PROJECT_ROOT = Path.cwd().parent

RAW_FILE = (
    PROJECT_ROOT
    / "data"
    / "raw"
    / "India_Flood_Inventory_v3 (2).csv"
)

print("Raw dataset path:")
print(RAW_FILE)

print("\nFile exists:", RAW_FILE.exists())

if not RAW_FILE.exists():
    raise FileNotFoundError(
        f"Raw India dataset not found:\n{RAW_FILE}"
    )

df = pd.read_csv(RAW_FILE)

print("\nRaw India dataset reloaded successfully.")
print("Shape:", df.shape)

print("\nColumns:")
for i, column in enumerate(df.columns, start=1):
    print(f"{i:>2}. {repr(column)}")
```

```python
# ============================================================
# TARGET LEAKAGE AUDIT — RAW INDIA DATASET
# ============================================================

print("=" * 100)
print("TARGET LEAKAGE AUDIT — RAW INDIA DATASET")
print("=" * 100)

TARGET = "Human fatality"

# Work only with the raw dataset
audit_df = df.copy()

# Convert target to numeric for analysis
audit_df[TARGET] = pd.to_numeric(
    audit_df[TARGET],
    errors="coerce"
)

# Create the target exactly as the current pipeline does
audit_df["High_Impact"] = (
    audit_df[TARGET] >= 10
).astype("Int64")

print("\nTarget definition:")
print("High_Impact = 1  -> Human fatality >= 10")
print("High_Impact = 0  -> Human fatality < 10")

print("\nRows with known target:")
print(audit_df[TARGET].notna().sum())

print("\nTarget distribution among rows with known fatality:")

print(
    audit_df.loc[
        audit_df[TARGET].notna(),
        "High_Impact"
    ].value_counts(dropna=False).sort_index()
)

print("\n" + "=" * 100)
print("POTENTIAL OUTCOME / LEAKAGE COLUMNS")
print("=" * 100)

potential_columns = [
    "Human fatality",
    "Human injured",
    "Human Displaced",
    "Animal Fatality",
    "Description of Casualties/injured",
    "Extent of damage",
    "Duration(Days)",
    "Area Affected"
]

for column in potential_columns:

    if column not in audit_df.columns:
        print(f"\n{column}: NOT FOUND")
        continue

    print("\n" + "-" * 80)
    print(f"COLUMN: {repr(column)}")

    print("dtype:", audit_df[column].dtype)
    print("non-null:", audit_df[column].notna().sum())
    print("missing:", audit_df[column].isna().sum())
    print(
        "missing %:",
        round(audit_df[column].isna().mean() * 100, 2)
    )
    print("unique non-null:", audit_df[column].nunique(dropna=True))

    print("\nSample non-null values:")

    values = (
        audit_df[column]
        .dropna()
        .head(10)
        .tolist()
    )

    for value in values:
        print(" ", repr(value))
```

```python
# ============================================================
# COMPACT LEAKAGE AUDIT SUMMARY
# ============================================================

audit_columns = [
    "Human fatality",
    "Human injured",
    "Human Displaced",
    "Animal Fatality",
    "Description of Casualties/injured",
    "Extent of damage",
    "Duration(Days)",
    "Area Affected"
]

rows = []

for column in audit_columns:

    if column not in df.columns:
        continue

    series = df[column]

    rows.append({
        "Column": repr(column),
        "Dtype": str(series.dtype),
        "Non_Null": int(series.notna().sum()),
        "Missing": int(series.isna().sum()),
        "Missing_%": round(series.isna().mean() * 100, 2),
        "Unique": int(series.nunique(dropna=True)),
        "Example_1": str(series.dropna().iloc[0])
        if series.notna().any() else "",
        "Example_2": str(series.dropna().iloc[1])
        if series.notna().sum() > 1 else "",
        "Example_3": str(series.dropna().iloc[2])
        if series.notna().sum() > 2 else ""
    })

leakage_summary = pd.DataFrame(rows)

display(leakage_summary)
```

```python
# ============================================================
# CASUALTY DESCRIPTION — TARGET LEAKAGE INVESTIGATION
# ============================================================

print("=" * 100)
print("CASUALTY DESCRIPTION — TARGET LEAKAGE INVESTIGATION")
print("=" * 100)

TEXT_COLUMN = "Description of Casualties/injured"
TARGET_COLUMN = "Human fatality"

if TEXT_COLUMN not in audit_df.columns:
    raise ValueError(f"Column '{TEXT_COLUMN}' not found.")

# Work from audit_df because it already contains High_Impact
analysis_df = audit_df.copy()

analysis_df[TEXT_COLUMN] = (
    analysis_df[TEXT_COLUMN]
    .astype("string")
    .str.lower()
    .str.strip()
)

print(
    "\nRows with known Human fatality:",
    len(analysis_df)
)

# ------------------------------------------------------------
# FATALITY-RELATED TERMS
# ------------------------------------------------------------

terms = [
    "death",
    "deaths",
    "died",
    "dead",
    "fatality",
    "fatalities",
    "killed",
    "casualties",
    "life lost",
    "lives lost",
    "washed away"
]

results = []

for term in terms:

    mask = (
        analysis_df[TEXT_COLUMN]
        .fillna("")
        .str.contains(term, regex=False)
    )

    results.append({
        "Term": term,
        "Rows_Containing_Term": int(mask.sum()),
        "High_Impact_Rows": int(
            analysis_df.loc[mask, "High_Impact"].sum()
        ),
        "Low_Impact_Rows": int(
            mask.sum()
            - analysis_df.loc[mask, "High_Impact"].sum()
        )
    })

results_df = pd.DataFrame(results)

print("\nFatality-related terminology frequency:")
print("-" * 100)

display(results_df)

# ------------------------------------------------------------
# REPRESENTATIVE MATCHING DESCRIPTIONS
# ------------------------------------------------------------

combined_mask = (
    analysis_df[TEXT_COLUMN]
    .fillna("")
    .apply(
        lambda text: any(
            term in text
            for term in terms
        )
    )
)

matching = analysis_df.loc[
    combined_mask
].copy()

print("\n")
print("=" * 100)
print("REPRESENTATIVE DESCRIPTIONS CONTAINING FATALITY TERMS")
print("=" * 100)

print(
    "\nNumber of matching rows:",
    len(matching)
)

if len(matching) > 0:

    display(
        matching[
            [
                TARGET_COLUMN,
                "High_Impact",
                TEXT_COLUMN
            ]
        ]
        .head(20)
    )
```

```python
# ============================================================
# EXTENT OF DAMAGE — TARGET LEAKAGE INVESTIGATION
# ============================================================

print("=" * 100)
print("EXTENT OF DAMAGE — TARGET LEAKAGE INVESTIGATION")
print("=" * 100)

TEXT_COLUMN = "Extent of damage "
TARGET_COLUMN = "Human fatality"

if TEXT_COLUMN not in audit_df.columns:
    raise ValueError(f"Column '{TEXT_COLUMN}' not found.")

analysis_df = audit_df.copy()

analysis_df[TEXT_COLUMN] = (
    analysis_df[TEXT_COLUMN]
    .astype("string")
    .str.lower()
    .str.strip()
)

print(
    "\nRows with known Human fatality:",
    len(analysis_df)
)

# ------------------------------------------------------------
# FATALITY-RELATED TERMS
# ------------------------------------------------------------

terms = [
    "death",
    "deaths",
    "died",
    "dead",
    "fatality",
    "fatalities",
    "killed",
    "casualties",
    "life lost",
    "lives lost",
    "people lost",
    "persons lost",
    "people died",
    "persons died",
    "people killed",
    "persons killed",
    "washed away"
]

results = []

for term in terms:

    mask = (
        analysis_df[TEXT_COLUMN]
        .fillna("")
        .str.contains(term, regex=False)
    )

    total_count = int(mask.sum())

    high_count = int(
        analysis_df.loc[
            mask,
            "High_Impact"
        ].sum()
    )

    results.append({
        "Term": term,
        "Rows_Containing_Term": total_count,
        "High_Impact_Rows": high_count,
        "Low_Impact_Rows": total_count - high_count
    })

results_df = pd.DataFrame(results)

print("\nFatality-related terminology frequency:")
print("-" * 100)

display(results_df)

# ------------------------------------------------------------
# REPRESENTATIVE MATCHING DESCRIPTIONS
# ------------------------------------------------------------

combined_mask = (
    analysis_df[TEXT_COLUMN]
    .fillna("")
    .apply(
        lambda text: any(
            term in text
            for term in terms
        )
    )
)

matching = analysis_df.loc[
    combined_mask
].copy()

print("\n")
print("=" * 100)
print("REPRESENTATIVE EXTENT-OF-DAMAGE DESCRIPTIONS")
print("=" * 100)

print(
    "\nNumber of matching rows:",
    len(matching)
)

if len(matching) > 0:

    display(
        matching[
            [
                TARGET_COLUMN,
                "High_Impact",
                TEXT_COLUMN
            ]
        ]
        .head(30)
    )
```

```python
# ============================================================
# FINAL OUTCOME / LEAKAGE COLUMN AUDIT
# ============================================================

print("=" * 100)
print("FINAL OUTCOME / LEAKAGE COLUMN AUDIT")
print("=" * 100)

OUTCOME_COLUMNS = [
    "Human fatality",
    "Human injured",
    "Human Displaced",
    "Animal Fatality",
    "Description of Casualties/injured",
    "Extent of damage",
    "Area Affected"
]

print("\nColumns identified as outcome/post-event information:")
print("-" * 100)

for column in OUTCOME_COLUMNS:

    if column in audit_df.columns:

        print(
            f"EXCLUDE: {column}"
        )

    else:

        print(
            f"NOT FOUND: {column}"
        )

print("\n" + "=" * 100)
print("PROPOSED MODEL FEATURE COLUMNS")
print("=" * 100)

candidate_columns = [
    "Start Date",
    "End Date",
    "Duration(Days)",
    "Main Cause",
    "Districts",
    "State",
    "District_LGD_Codes",
    "State_Codes",
    "Event Source"
]

for column in candidate_columns:

    if column in audit_df.columns:

        print(f"KEEP CANDIDATE: {column}")

    else:

        print(f"NOT AVAILABLE: {column}")
```
