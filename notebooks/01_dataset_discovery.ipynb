```python
import re


def contains_keyword(column_name, keywords):
    """
    Check whether a column name contains a meaningful keyword
    rather than matching arbitrary character substrings.
    """
    name = str(column_name).lower()

    # Split column names such as:
    # Station_Number -> ["station", "number"]
    # Relative_Humidity -> ["relative", "humidity"]
    tokens = re.split(r"[^a-z0-9]+", name)

    return any(
        keyword in tokens
        for keyword in keywords
    )


def classify_columns(df):
    numerical = df.select_dtypes(include="number").columns.tolist()
    categorical = df.select_dtypes(
        include=["object", "string", "category"]
    ).columns.tolist()

    datetime_candidates = []
    geographic_candidates = []
    identifier_candidates = []
    target_candidates = []

    datetime_keywords = [
        "date", "time", "timestamp",
        "year", "month", "day", "period"
    ]

    geographic_keywords = [
        "latitude", "longitude",
        "lat", "lon",
        "altitude", "location",
        "station", "district",
        "state", "city", "region",
        "country", "coordinate", "coordinates"
    ]

    identifier_keywords = [
        "id", "identifier",
        "serial", "number",
        "code"
    ]

    target_keywords = [
        "target", "label",
        "class", "outcome",
        "flood", "disaster",
        "event"
    ]

    for column in df.columns:

        if contains_keyword(column, datetime_keywords):
            datetime_candidates.append(column)

        if contains_keyword(column, geographic_keywords):
            geographic_candidates.append(column)

        if contains_keyword(column, identifier_keywords):
            identifier_candidates.append(column)

        if contains_keyword(column, target_keywords):
            target_candidates.append(column)

    return {
        "numerical": numerical,
        "categorical": categorical,
        "datetime_candidates": datetime_candidates,
        "geographic_candidates": geographic_candidates,
        "identifier_candidates": identifier_candidates,
        "target_candidates": target_candidates
    }


for file in csv_files:
    df = pd.read_csv(file)

    classification = classify_columns(df)

    print("=" * 80)
    print(f"Dataset: {file.relative_to(PROJECT_ROOT)}")

    print("\nNumerical:")
    print(classification["numerical"])

    print("\nCategorical:")
    print(classification["categorical"])

    print("\nPotential datetime:")
    print(classification["datetime_candidates"])

    print("\nPotential geographic:")
    print(classification["geographic_candidates"])

    print("\nPotential identifier:")
    print(classification["identifier_candidates"])

    print("\nPotential target:")
    print(classification["target_candidates"])
```

```python
column_profiles = []

for file in csv_files:
    df = pd.read_csv(file)

    classification = classify_columns(df)

    for column in df.columns:

        roles = []

        if column in classification["numerical"]:
            roles.append("numerical")

        if column in classification["categorical"]:
            roles.append("categorical")

        if column in classification["datetime_candidates"]:
            roles.append("datetime_candidate")

        if column in classification["geographic_candidates"]:
            roles.append("geographic_candidate")

        if column in classification["identifier_candidates"]:
            roles.append("identifier_candidate")

        if column in classification["target_candidates"]:
            roles.append("target_candidate")

        column_profiles.append({
            "dataset": str(file.relative_to(PROJECT_ROOT)),
            "column": column,
            "dtype": str(df[column].dtype),
            "missing": int(df[column].isna().sum()),
            "unique_values": int(df[column].nunique(dropna=True)),
            "roles": ", ".join(roles)
        })

column_profile_df = pd.DataFrame(column_profiles)

column_profile_df
```

```python
target_candidates = column_profile_df[
    column_profile_df["roles"].str.contains(
        "target_candidate",
        na=False
    )
].copy()

target_candidates[
    [
        "dataset",
        "column",
        "dtype",
        "missing",
        "unique_values",
        "roles"
    ]
]
```
